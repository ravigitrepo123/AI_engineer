```python
x = 10  

def update_x():
    x += 5    

update_x()
print(x)  
```

```python
x = 10  

def update_x():
    global x
    x += 5    

update_x()
print(x)  
```

```python
# globals() 

x = 10  # Global variable

def update_x():
    globals()['x'] += 7

update_x()
print(x)  # Output: 15
```

```python
def count(lst): 
    
    even = 0
    odd = 0
    
    for i in lst:
        if i%2 == 0:
            even += 1
        else:
            odd +=1 
    return even,odd

lst = [10, 9, 8, 50, 34, 100] 
even, odd = count(lst)         

print(even)
print(odd)
```
